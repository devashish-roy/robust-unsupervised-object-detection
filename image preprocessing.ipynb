```python


# Define the folder containing your .png images
num_images = 89
input_folder = fr"C:\Users\devas\OneDrive - BTH Student\Documents\Dev\1. LTU\1. Master thesis\snowball\{num_images}\train-frames\png"
output_folder = fr"C:\Users\devas\OneDrive - BTH Student\Documents\Dev\1. LTU\1. Master thesis\snowball\{num_images}\train-frames\jpg"

# Ensure the output folder exists
os.makedirs(output_folder, exist_ok=True)
cnt = 0
# Loop through all files in the input folder
for filename in os.listdir(input_folder):
    #if filename.endswith(".png") and "2022-" in filename:  # Process only .png files
    if filename.endswith(".png"):  # Process only .png files
        cnt += 1
        png_image_path = os.path.join(input_folder, filename)

        #Extract the numeric part and format it to 5 digits
        #numeric_part = filename.split("frame")[-1].split(".png")[0]
        #formatted_numeric_part = f"{int(numeric_part):05d}"
            
        # New file name
        #new_filename = f"{formatted_numeric_part}.jpg"

        jpg_image_path = os.path.join(output_folder, filename.replace(".png", ".jpg"))        
        #jpg_image_path = os.path.join(output_folder, new_filename)

        # Open the .png image
        with Image.open(png_image_path) as img:
            # Convert image to RGB (JPEG doesn't support transparency)
            rgb_img = img.convert("RGB")
            # Save the image as .jpg
            rgb_img.save(jpg_image_path, "JPEG")
        if cnt % 500 == 0:
            print(f"Converted {filename} to {jpg_image_path}")

print("All images have been converted!")
```
