### Utility to create csv for ground truth
##### Prior to executing this, all the labels and the images in jpg format should be available.
##### ground_truth.csv is save in the tracking result folder of the exepriment folder

```python
import os
import pandas as pd
```

```python
def load_yolo_ground_truth(label_dir, image_dir):
    """
    label_dir: path to folder containing .txt files with YOLO labels
    image_dir: path to original images (for width & height)
    """
    import cv2

    gt_list = []
    cnt = 0

    for filename in os.listdir(label_dir):
        if filename.endswith(".txt"):
            original_image_name = f'{filename.replace(".txt", "")}.jpg'
            txt_path = os.path.join(label_dir, filename)
            img_path = os.path.join(image_dir, original_image_name)

            if not os.path.isfile(txt_path):
                print(f"File does not exist: {filename}")
                continue # skip if file doensot exist

            #Extract the numeric part and format it to 5 digits
            numeric_part = filename.split("frame")[-1].split(".txt")[0]
            frame_id = f"{int(numeric_part):05d}"

            # Get image shape
            img = cv2.imread(img_path)
            if img is None:
                print(f"Image does not exist: {original_image_name}")
                continue  # skip if image not found
            h, w = img.shape[:2]

            with open(txt_path, "r") as f:
                for line in f:
                    cls, x_c, y_c, bw, bh = map(float, line.strip().split())
                    x_min = (x_c - bw / 2) * w
                    y_min = (y_c - bh / 2) * h
                    x_max = (x_c + bw / 2) * w
                    y_max = (y_c + bh / 2) * h
                    gt_list.append({
                        "image_name": original_image_name,
                        "frame_id": frame_id,
                        "x_min": x_min,
                        "y_min": y_min,
                        "x_max": x_max,
                        "y_max": y_max,
                        "class_id": int(cls)
                    })

            # count
            cnt += 1

            # print the progress
            if cnt % 500 == 0:
                print(f'{cnt} images processed')

    return pd.DataFrame(gt_list)
```

```python
num_images = 4000
experiment_name = "finalPipeline-Sam2VideoImprov-Experiment"
PATH_TO_IMAGES = f"/home/devsin-3/1. Master thesis/snowball/{num_images}/train-frames/jpg"
PATH_TO_LABELS = f"/home/devsin-3/1. Master thesis/snowball/{num_images}/train-frames/labels"

ground_truth_df = load_yolo_ground_truth(PATH_TO_LABELS, PATH_TO_IMAGES)
ground_truth_df.to_csv(f"/home/devsin-3/1. Master thesis/snowball/{num_images}/{experiment_name}/tracking-results/ground_truth.csv", index=False, sep = ";")
```
